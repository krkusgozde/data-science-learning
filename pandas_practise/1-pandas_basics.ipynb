# 4. Eksik Veriler ve Veri Temizleme

```python
#isnull() >> eksik değerleri bulma
#isnull().sum() >> her sütundaki eksik değerleri sayma
#dropna() >> eksik değer içeren satırları silme
#fillna() >> eksik değerleri doldurma
```

```python
veriler_eksik = {
    "İsim": ["Ayşe", "Mehmet", "Zeynep", "Ali", "Ece"],
    "Yaş": [21, None, 22, 25, None],
    "Not": [85, 70, None, 60, 90]
}

df_eksik = pd.DataFrame(veriler_eksik)
```

```python
df_eksik
```

```python
df_eksik.isnull()
```

```python
df_eksik.isnull().sum()
```

```python
df_eksik.dropna()
```

```python
df_eksik.fillna(0)
```

```python
df_dolu =df_eksik["Yaş"].fillna(22)
```

```python
df_dolu
```

```python
df_eksik["Not"].fillna(df_eksik["Not"].mean())
```

```python
df_dolu = df_eksik.copy()
df_dolu["Yaş"] = df_dolu["Yaş"].fillna(22)
df_dolu["Not"] = df_dolu["Not"].fillna(df_dolu["Not"].mean())
```

# 5. Gruplama ve Özel İstatistikler

```python
calisanlar = {
    "İsim": ["Ayşe", "Mehmet", "Zeynep", "Ali", "Ece", "Can", "Elif"],
    "Departman": ["Veri", "İK", "Veri", "Finans", "İK", "Veri", "Finans"],
    "Maaş": [35000, 28000, 42000, 32000, 30000, 39000, 34000],
    "Deneyim": [2, 1, 4, 3, 2, 3, 2]
}

df_calisan = pd.DataFrame(calisanlar)
```

```python
df_calisan
```

```python
#value_counts >> bir sütundaki her farklı değerin kaç kez tekrarlandığını  sayar
```

```python
df_calisan["Departman"].value_counts()
```

```python
#groupby >> verileri belirli kategorilere göre gruplayıp
#özet istatistikler (ortalama, toplam, sayım vb.) çıkarmaya yarar.
```

```python
df_calisan.groupby("Departman")["Maaş"].mean()
```

```python
df_calisan.groupby("Departman")["Maaş"].max()
```

```python
df_calisan.groupby("Departman")["Maaş"].min()
```

```python
# agg() >> birden fazla istatistiği tek seferde hesaplar
```

```python
df_calisan.groupby("Departman")["Maaş"].agg("mean" , "max")
```

```python
df_calisan.groupby("Departman")["Deneyim"].mean()
```

```python
df_calisan[df_calisan["Maaş"] > 35000]
```

# 6. Tabloları Birleştirme

```python
#merge() >> ortak bilgiye göre birleştirme yapar
#concat() >> genellikle aynı sütunlara sahip tabloları alt alta veya yan yana eklemek için kullanılır
```

```python
# Birinci tablo: Öğrenci bilgileri
ogrenciler = {
    "Ogrenci_ID": [101, 102, 103, 104],
    "İsim": ["Ayşe", "Mehmet", "Zeynep", "Ali"],
    "Bölüm": ["Matematik", "Fizik", "Matematik", "Kimya"]
}

df_ogrenci = pd.DataFrame(ogrenciler)
```

```python
notlar = {
    "Ogrenci_ID": [101, 102, 103, 104],
    "Vize": [80, 65, 90, 70],
    "Final": [90, 75, 95, 60]
}

df_notlar = pd.DataFrame(notlar)
```

```python
df_ogrenci
```

```python
df_notlar
```

```python
pd.merge(df_ogrenci, df_notlar  , on = "Ogrenci_ID" )
```

```python
sinif_a = pd.DataFrame({
    "İsim": ["Elif", "Can"],
    "Not": [85, 75]
})

sinif_b = pd.DataFrame({
    "İsim": ["Deniz", "Ece"],
    "Not": [90, 80]
})
```

```python
pd.concat([sinif_a,sinif_b] , ignore_index= True) #ignore_index >> eski indeks numaralarını yok sayıp yeni sıralı indeksler oluşturmak için kullanılır
```

```python
sinif_c = pd.DataFrame({
    "İsim": ["Merve", "Ahmet"],
    "Not": [88, 72]
})

sinif_d = pd.DataFrame({
    "İsim": ["Burak", "Selin"],
    "Not": [65, 95]
})
```

```python
pd.concat([sinif_c , sinif_d] ,ignore_index = True)
```

```python
ogrenciler = pd.DataFrame({
    "Ogrenci_ID": [1, 2, 3],
    "İsim": ["Ayşe", "Mehmet", "Zeynep"]
})

sinav_sonuclari = pd.DataFrame({
    "Ogrenci_ID": [1, 2, 3],
    "Not": [85, 70, 95]
})
```

```python
pd.merge(ogrenciler,sinav_sonuclari , on = "Ogrenci_ID")
```
