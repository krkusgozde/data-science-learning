```python
#Bir while döngüsü kur.  
#Döngü kullanıcıdan kelime istesin.
#Kullanıcı "çıkış" yazana kadar alınan kelimeleri ekrana yazdırsın.
#Kullanıcı "çıkış" yazdığında break ile döngüyü bitirip ekrana "Sistemden çıkıldı" yazsın.
```

```python
while True:
    kelime = input("bir kelime yazın (çıkmak için 'çıkış' yazın): ")

    if kelime == "çıkış": 
        print("sistemden çıkıldı. ")
        break

    print(f"girdiğiniz kelime : {kelime}")
```

```python
#Sayı Toplama Uygulaması yapalım. 
#Kullanıcı sayı girdikçe sayıları toplayacağız
#durmak istediğinde sistemi kapatacağız.
```

```python

toplam = 0

while True:
    girdi = input("Bir sayı girin (Çıkmak için 'dur' yazın): ")
    
    if girdi == "dur":
        print("Toplama sona erdi.")
        break
    else:

        sayi = int(girdi)
        toplam = toplam + sayi
        print(f"Güncel Toplam: {toplam}")

print(f"Sonuç: {toplam}")
    
    
            
             
    
             
```

```python
#Kullanıcıdan input() ile yaşını alıp bunu bir sayıya (int) çevir.
#Eğer yaşı 18'den küçükse ekrana "Ehliyet alamazsınız" yazdır.
#Değilse (else) ekrana "Ehliyet alabilirsiniz" yazdır.
```

```python
age = int(input("Lutfen yasinizi girin: "))

if age >= 18:
    print("Ehliyet alabilirsiniz")
else:
    print("Ehliyet alamazsınız!")
```

```python
#Sayı Sayma Aparatı
#Kullanıcıdan pozitif bir sayı alacağız ve 1'den başlayarak
#o sayıya kadar (o sayı dahil) olan sayıları ekrana sırayla yazdıracağız.
```

```python
hedef = int(input("Lutfen sayiyi giriniz: "))
sayac = 1

while sayac <= hedef:
    print(sayac)
    sayac = sayac + 1
```

```python
#Roket Fırlatma Geri Sayımı 
#Kullanıcıdan başlama sayısını alacağız ve 
#1'e kadar geriye doğru saydırıp döngü bitince "Fırlatma Başladı!" yazdıracağız.
```

```python
sayac = int(input("Kactan geriye sayayim?: "))
while sayac >= 1:
    print(sayac)
    sayac = sayac - 1

print("Fırlatma Basladi!")
```

```python
#Çift Sayı Filtresi
#Kullanıcı bir sayı girecek. 1'den başlayıp kullanıcının girdiği sayıya kadar sayacağız.
#Ama bu sefer ekrana SADECE ÇİFT SAYILARI yazdıracağız.
```

```python
hedef = int(input("Lutfen bir sayi giriniz:"))
sayac = 1
while sayac <= hedef:
    if sayac % 2 == 0:
        print(sayac)
    sayac = sayac + 1    
```
