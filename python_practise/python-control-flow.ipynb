```python
#Bir for döngüsü kullanarak 1'den 20'ye kadar (20 dahil) olan sayıları sırayla inceleyen bir Python kodu yazın.
#Kodunuz şu mantıkla çalışmalı: 
#Eğer sayı 3'ün katı ise (örneğin 3, 6, 9...), ekrana "X sayısı 3'ün katı olduğu için atlandı" yazdırın ve bu sayıyı işleme almadan pas geçin. İşleme alınan (atlanmayan) sayıların toplamı 30'u veya daha fazlasını geçerse, ekrana "Toplam 30 sınırını aştı, döngü durduruluyor!" yazdırın ve döngüyü anında bitirin.
#Ekrana Yazdırma: Yukarıdaki iki duruma da uymayan sayıları ekrana yazdırın ve anlık toplamı gösterin.
```

```python
toplam = 0

for sayi in range(1,21):
    

    if sayi %3 == 0:
        print(f"{sayi} sayısı 3 ' ün katı olduğu için atlandı!!")
        continue
    toplam = toplam + sayi
    print(f"{sayi} | Anlık Toplam: {toplam}")
    
    if toplam >= 30:
        print("Durduruldu! ")
        break
        
```

```python
#Senaryo: "Alışveriş Sepeti Taraması"
#Bir mağaza kasasında olduğumuzu düşün. Elimizde şu ürün fiyatlarının olduğu bir liste var:
#fiyatlar = [15, 0, 45, -10, 80, 200, 30]
#Senin görevin bir for döngüsü kurup bu liste içindeki fiyatları sırayla kontrol etmek:
#Hatalı Ürün / Hediye (continue): Eğer ürün fiyatı 0 veya 0'dan küçük (negatif) ise, ekrana "Hatalı ürün atlandı!" yazdır ve bu fiyatı toplama eklemeden pas geç.
#Kasa Limiti (break): Fiyatı geçerli olan ürünleri bir toplam_tutar değişkenine ekle. Eğer toplam_tutar 100 TL veya üstüne ulaşırsa, ekrana "Kasa limiti aşıldı, işlem sonlandırılıyor!" yazdır ve döngüyü anında bitir.
#Geçerli Ürün: Yukarıdaki iki duruma da uymayan ürünler için ekrana ürünün fiyatını ve o anki toplam tutarı yazdır.
```

```python
fiyat_listesi = [15,0,45,-10,80,200,30]
toplam_tutar = 0

for fiyat in fiyat_listesi:

    if fiyat <= 0:
        print(f"{fiyat} Hatalı ürün atlandı!!")
        continue
    toplam_tutar = toplam_tutar + fiyat     

    if fiyat > 0 :
        print(f" Eklendi : {fiyat} TL | Anlık Toplam: {toplam_tutar} TL")

    if toplam_tutar > 100:
        print("kasa limiti aşıldı,işlem sonlandırılıyor! ")
        break
    
```

```python
#Senaryo: "Sayısal Loto / Şanslı Sayı Arama"
#Elimizde bilgisayarın ürettiği/belirlediği 10 adet şanslı sayıdan oluşan bir liste var:
#sansli_sayilar = [12, 45, 8, 23, 19, 34, 50, 3, 29, 42]
#Kurallar:
#Çift ve 20'den Küçük Sayılar (continue): Eğer sayı hem ÇİFT hem de 20'den KÜÇÜKSE (örneğin 12 veya 8), ekrana "{sayı} - Pas geçildi (Çift ve < 20)" yazdır ve bu sayıyı işleme almadan pas geç (continue).
#Hedef Sayı / Büyük İkramiye (break): Eğer sayı 50 ise, ekrana "Hedef sayı 50 bulundu! Arama durduruluyor." yazdır ve döngüyü anında bitir (break).
#Diğer Sayılar: Yukarıdaki kurallara takılmayan sayıları ekrana "{sayı} - İncelemeye alındı" şeklinde yazdır.
```

```python
sansli_sayilar = [12,45,8,23,19,34,50,3,29,42]

for sayi in sansli_sayilar:
    
    if sayi % 2 == 0 and sayi < 20:
        print(f"{sayi} Pas geçildi!! ")
        continue
    if sayi == 50:
        print("Hedef sayı bulundu, arama durduruluyor!!")
        break
    else:
        print(f"{sayi} İncelemeye alındı")
```

```python
#Senaryo: "Fabrika Üretim Kalite Kontrolü"
#Bir fabrikada banttan çıkan ürünlerin ağırlıklarını kontrol ediyoruz:
#agirliklar = [150, 45, 200, 300, 80, 220, 500, 180]
#Bir for döngüsü kurarak bu ürünleri sırayla inceleyeceksin. Ayrıca hatalı ürün sayısını tutmak için döngünün dışında hatali_urun_sayisi = 0 şeklinde bir sayaç oluşturmalısın.
#Kurallar:
#Standart Altı Ürün (continue): Eğer ürün ağırlığı 100 gramdan azsa (örneğin 45 veya 80), ekrana "{agirlik}g - Standart altı, atlandı!" yazdır. Hatalı ürün sayacını 1 artır ve bu ürünü pas geç (continue).
#Kritik Hata / Bant Durdurma (break): Eğer toplam hatali_urun_sayisi 2'ye ulaşırsa, ekrana "Üst üste 2 hatalı ürün! Üretim bandı durduruluyor!" yazdır ve döngüyü anında bitir (break).
#Kabul Edilen Ürün: Yukarıdaki kurallara takılmayan ürünler için ekrana "{agirlik}g - Kalite kontrolden geçti." yazdır.
```

```python
agirliklar = [150,45,200,300,80,220,500,180]
hatali_urun_sayisi = 0

for agirlik in agirliklar:

    if agirlik < 100:
        print(f"{agirlik}g Standart altı, atlandı!!")
        hatali_urun_sayisi = hatali_urun_sayisi + 1
        
    if hatali_urun_sayisi == 2:
        print("Üst üste 2 hatali urun!! Üretim bandı durduruluyor")
        break
    else:
        print(f"{agirlik}g Kaliteli ürün,kontrolden geçti!")
        continue
        
```
