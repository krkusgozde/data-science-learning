```python
#Güvenli Liste Elemanı Çekme
#Bu pratikte, listede olmayan bir indekse (sıraya) erişmeye çalıştığımızda
#çıkan IndexError hatasını ve harf girildiğinde çıkan ValueError hatasını yakalayacağız.
```

```python
def eleman_getir(liste, indeks):
    try:
        indeks = int(indeks)
        return liste[indeks]
    except IndexError:
        return "Hata: Aradığınız indeks listede bulunamadı!"

    except ValueError:
        return "Hata: İndeks bir sayı olmalıdır!"

meyveler = ["elma" , "armut" , "muz" , "çilek"]

print(eleman_getir(meyveler,1))
print(eleman_getir(meyveler, 10))
print(eleman_getir(meyveler, "iki"))
        
```

```python
#Güvenli Sözlük Verisi Çekme ve Sözlük İçi Hata Yönetimi
#Bir e-ticaret sisteminde ürün fiyatlarını tutan bir sözlüğümüz var. Kullanıcı bir ürün adı girecek ve biz fiyatını döndüreceğiz.
#İstekler:
#fiyat_sorgula adında bir fonksiyon yaz.
#Fonksiyon iki parametre alsın: urun_fiyatlari (sözlük) ve urun_adi (metin).
#Fonksiyon sözlükte aranan ürün varsa fiyatını döndürsün.
#Hata Yönetimi (try-except):
#Eğer aranan ürün sözlükte yoksa Python normalde KeyError hatası verir.
#try-except kullanarak KeyError hatasını yakala ve ekrana "Hata: Aradığınız ürün stokta/sistemde bulunamadı!" mesajı döndür.
```

```python
def fiyat_sorgula(urun_fiyatlari, urun_adi):
    try:
        return urun_fiyatlari[urun_adi]
    except KeyError:
        return "Hata: Aradığınız ürün stokta/sistemde bulunamamaktadır!"

urunler = {"laptop": 25000 , "telefon": 15000 , "kulaklik": 2000}

sonuc1 = fiyat_sorgula(urunler, "laptop")
print("laptop fiyatu:" , sonuc1)

sonuc2 = fiyat_sorgula(urunler, "masaustu")
print("masaustu durumu: " , sonuc2)

```

```python
#Güvenli Sayı Dönüştürme
#Soru:
#sayiya_cevir adında bir fonksiyon yaz.
#Dışarıdan bir veri parametresi alsın.
#try içinde bu veriyi int() kullanarak tam sayıya çevirmeyi denesin ve çevrilmiş halini döndürsün (return).
#Eğer gelen veri sayıya çevrilemiyorsa (ValueError oluşursa), program çökmesin; "Hata: Girdiğiniz değer bir tam sayıya dönüştürülemez!" mesajı döndürsün.
```

```python
def sayiya_cevir(sayi):
    try:
        return int(sayi)
    except ValueError:
        return "Hata: Girdiğiniz değer bir tam sayıya dönüştürülemez!!"

print(sayiya_cevir(123))
print(sayiya_cevir("gozde"))
```

```python
#Güvenli Bölme ve Tip Kontrolü
#Soru:
#guvenli_bolme adında bir fonksiyon yaz.
#İki parametre alsın: sayi1 ve sayi2.
#try bloğu içinde bu iki sayıyı birbirine bölsün (sayi1 / sayi2) ve sonucu döndürsün.
#Birden fazla hata yakala:
#sayi2 değeri 0 girilirse (ZeroDivisionError): "Hata: Bir sayı sıfıra bölünemez!" dönsün.
#Parametrelerden biri sayı yerine metin girilirse (TypeError): "Hata: Lütfen sadece sayısal değerler girin!" dönsün.
```

```python
def guvenli_bolme(sayi1,sayi2):
    try:
        return sayi1 / sayi2
    except ZeroDivisionError:
        return "Hata: Bir sayı sıfıra bölünemez!"

    except TypeError:
        return("Hata: Lütfen sayısal değerler girin! ")

print(guvenli_bolme(10,2))
print(guvenli_bolme(10,0))
print(guvenli_bolme(10,"g"))
```

```python
#Liste ve Döngü İçinde Hata Yönetimi
#Soru:
#listeyi_temizle adında bir fonksiyon yaz.
#Dışarıdan karışık elemanlardan oluşan bir liste alsın (örn: [10, "5", "abc", 20, "30", None]).
#Bir temiz_liste adında boş liste oluştur.
#Bir for döngüsü ile listedeki her elemanı gez.
#try bloğu içinde elemanı int() ile sayıya çevirip temiz_listeye ekle (append).
#Eğer eleman sayıya çevrilemiyorsa (ValueError veya TypeError alıyorsa), program çökmesin, o elemanı atlayıp (pass veya boş geçerek) döngüye devam etsin.
#Fonksiyonun sonunda sadece başarıyla sayıya dönüştürülmüş elemanların olduğu listeyi döndürsün.
```

```python
def listeyi_temizle(liste):
    temiz_liste = []
    for eleman in liste:
        try:
            sayi = int(eleman)
            temiz_liste.append(sayi)
        except(ValueError, TypeError):
             pass
    return temiz_liste

karisik_liste = [10,"5", "abc" , 20, "30" , None, "100"]
print(listeyi_temizle(karisik_liste))
```
