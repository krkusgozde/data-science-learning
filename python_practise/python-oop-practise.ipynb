```python
#Banka Hesabı Sınıfı
#Soru:
#Bir banka hesabı şablonu oluşturacağız.
#BankaHesabi adında bir sınıf (class) oluştur.
#__init__ metodu 2 parametre alsın: hesap_sahibi ve bakiye. Bunları self.hesap_sahibi ve self.bakiye olarak atansın.
#Sınıfın içine bilgi_ver adında bir metot yaz. Bu metot: "Hesap Sahibi: [hesap_sahibi], Bakiye: [bakiye] TL" metnini döndürsün (return).
```

```python
class BankaHesabi:
    def __init__(self,hesap_sahibi,bakiye):
        self.hesap_sahibi = hesap_sahibi
        self.bakiye = bakiye

    def bilgi_ver(self):
        return f"Hesap Sahibi: {self.hesap_sahibi}, Bakiye: {self.bakiye} TL "

hesap = BankaHesabi("gozde" ,10000)
print(hesap.bilgi_ver())
```

```python
#Bakiye Güncelleyen Metotlar (Para Yatır / Çek)
#Soru:
#BankaHesabi sınıfını geliştireceğiz. Bir önceki sınıfına 2 yeni davranış (metot) ekle:
#para_yatir(self, miktar): self.bakiye değerine gelen miktarı eklesin ve "Yeni bakiye: [bakiye] TL" döndürsün.
#para_cek(self, miktar):
#Eğer çekilmek istenen miktar bakiyeden büyükse: "Yetersiz bakiye!" döndürsün.
#Bakiyeden küçük veya eşitse: self.bakiye değerinden düşsün ve "Kalan bakiye: [bakiye] TL" döndürsün.
```

```python
class BankaHesabi:
    def __init__(self,hesap_sahibi,bakiye):
        self.hesap_sahibi = hesap_sahibi
        self.bakiye = bakiye 

    def para_yatir(self,miktar):
        self.bakiye = self.bakiye + miktar
        return f"Yeni bakiye: {self.bakiye} TL"

    def para_cek(self,miktar):
        if miktar > self.bakiye:
            return "Yetersiz Bakiye!!"

        else:
            self.bakiye = self.bakiye - miktar
            return f"Kalan Bakiye: {self.bakiye} TL"

hesap = BankaHesabi("gozde", 10000)

print(hesap.para_yatir(500))
print(hesap.para_cek(20000))
print(hesap.para_cek(7000))
```

```python
#Soru:
#Geliştirdiğin BankaHesabi sınıfındaki para_cek metodunu güvenli hale getireceğiz.
#para_cek(self, miktar) metodunda miktar değerinin sayısal bir veri olup olmadığını denetle.
#Metot içinde try-except bloğu kullan:
#float(miktar) ile gelen parametreyi sayıya çevirmeyi dene.
#Eğer hatalı bir veri girildiyse (ValueError alırsan): "Hata: Lütfen geçerli bir sayısal miktar girin!" döndürsün.
#Dönüştürme başarılı olursa normal bakiye kontrolü ve para çekme işlemini gerçekleştirsin.
```

```python
class BankaHesabi:
    def __init__(self,hesap_sahibi,bakiye):
        self.hesap_sahibi = hesap_sahibi
        self.bakiye = bakiye

    def para_cek(self,miktar):
        try:
            miktar = float(miktar)
        except ValueError:
            return "Hata: Lütfen geçerli bir sayısal miktar girin!"

        if miktar > self.bakiye:
            return "Yetersiz bakiye!"

        else:
            self.bakiye -= miktar
            return f"Kalan Bakiye: {self.bakiye} TL"

hesap = BankaHesabi("gozde", 1000)
print(hesap.para_cek("yüz"))
print(hesap.para_cek(200))

        
```

```python
#Bir Ogrenci sınıfı oluşturacağız.
#İstekler:
#__init__ içinde ad, soyad ve notlar (boş bir liste []) özniteliklerini (attribute) tanımla.
#not_ekle(self, yeni_not) adında bir metot yaz. Gelen notu notlar listesine eklesin.
#ortalama_hesapla(self) adında bir metot yaz. Liste boşsa 0 dönsün, değilse notların ortalamasını hesaplayıp döndürsün.     
    
        
```

```python
class Ogrenci:
    def __init__(self,ad,soyad):
        self.ad = ad
        self.soyad = soyad
        self.notlar = []

    def not_ekle(self,yeni_not):
        self.notlar.append(yeni_not)

    def ortalama_hesapla(self):
        if not self.notlar:
            return 0
        return sum(self.notlar) / len(self.notlar)


ogrenci1 = Ogrenci("gozde" , "karakus")
ogrenci1.not_ekle(95)
ogrenci1.not_ekle(88)
print(f"{ogrenci1.ad} {ogrenci1.soyad} Ortalaması:" , ogrenci1.ortalama_hesapla())
         
```
