```python
hava_sicakligi = int(input("Lutfen hava sicakliginiz girin:"))

if hava_sicakligi > 30:
    print("Çok sıcak, tişört giyin!")
elif hava_sicakligi >= 15 :
    print("Hava güzel, hırka alabilirsiniz. ")
elif hava_sicakligi >= 14:
    print("Hava soğuk, mont giyin")
else:
    print("Dondurucu soğuk, dışarı çıkmayın!")
                     
```

```python
#Sistemde tanımlı bilgiler şunlardır:
#Kullanıcı adı: "admin"
#Şifre: "1234"
#Kullanıcıdan kullanıcı adını ve şifresini alıp sistemi kontrol edeceksin.
```

```python
kullanici_adi = input("Lutfen kullanici adinizi giriniz: ")
sifre = input("Lutfen sifrenizi girin: ")

if kullanici_adi == "admin" and sifre == "1234":
    print("Giriş başarılı! ")
elif kullanici_adi == "admin" and sifre != "1234":
    print("Şifre hatalı!")
elif kullanici_adi != "admin" and sifre == "1234":
    print("Kullanıcı adı hatalı!")
else:
    print("Kullanıcı adı ve şifre hatalı! ")
    

```

```python
#Kullanıcıdan bir tam sayı alacaksın ve
#bu sayının 3'e veya 5'e bölünüp bölünmediğini kontrol edeceksin.
```

```python
sayi = int(input("Lutfen bir sayi giriniz: "))
if sayi %3 == 0 and sayi %5 == 0:
    print("Sayı 15'in katıdır 'hem 3 e hem 5 e bölünür ")
elif sayi % 3 == 0 and sayi % 5 !=0 :
    print("Sayı 3' ün katıdır")
elif sayi % 3 != 0 and sayi % 5 == 0:
    print("Sayı 5'in katıdır")
else:
    print("Sayı ne 3' e ne de 5' e bölünür")
    
```

```python
#Sıcaklık Birim Dönüştürücü
#Kullanıcıdan dönüştürmek istediği sıcaklık değerini (sayı) ve hedef birimi alacaksın.
```

```python
sicaklik = float(input("Lutfen sicaklik degerini giriniz: "))
secim = input("Hangi birime dönüştürülsün? (Fahrenheit için 'F', Celsius için ('C'): ")


if secim == 'F' or secim == 'f':
    sonuc = (sicaklik * 1.8) + 32
    print(f"Fahrenheit Değeri: {sonuc}")
elif secim == 'C'  or secim == 'c':
    sonuc = (sicaklik - 32 ) / 1.8
    print(f"Celsius Değeri: {sonuc}")
else:
    print("Geçersiz seçim yapıldı, lütfen 'C' veya 'F' girin")
```
