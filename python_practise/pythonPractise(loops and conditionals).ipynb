```python
#Kullanıcıdan bir sayı al ve 1'den girilen 
#sayıya kadar (girilen sayı dahil) 
#olan tüm sayıları ve karelerini ekrana yazdır.
```

```python
sayi = int(input("Lutfen bir sayi giriniz: "))

for i in range(1, sayi + 1):
    karesi = i ** 2
    print(f"{i} sayısının karesi: {karesi}")
```

```python
#Kullanıcıdan yine bir sayı al. 
#for döngüsüyle 1'den o sayıya kadar dön. 
#Ancak sadece ÇİFT sayıların karesini ekrana yazdır!
```

```python
sayi = int(input("lutfen bir sayı giriniz: "))

for i in range(1, sayi + 1):
    if i % 2 == 0:
        print(f"{i} sayısının karesi: {i ** 2}")
```

```python
#Görev: Kullanıcıdan doğru şifreyi girene kadar sürekli şifre isteyen bir program yazınız.

#Kurallar:
#Doğru şifre "python123" olsun.
#Kullanıcı yanlış girdiği sürece program "Hatalı şifre, tekrar deneyin!" desin ve tekrar şifre sorsun.
#Doğru girdiğinde döngü bitsin ve ekrana "Giriş başarılı! Hoş geldiniz." yazdırsın.
```

```python
dogru_sifre = "python123"
girilen_sifre = " "

while girilen_sifre != dogru_sifre:
    girilen_sifre = input("Lutfen sifreyi giriniz: ")
    if girilen_sifre == dogru_sifre:
        print("Giriş başarılı!")
    else:
        print("Hatalı şifre girdiniz! ")
    
```

```python
#Görev: Kullanıcının 3 hakkı olan bir Sayı Tahmin Oyunu yazınız.
```

```python
gizli_sayi = 7
hak = 3

while hak > 0 :
    tahmin = int(input("tahmininizi girin: "))
    if tahmin == gizli_sayi:
        print("Tebrikler doğru bildiniz ")
        break
    else:
        hak = hak - 1
        print(f"Yanlış tahmin! Kalan hakkınız {hak}")
    
    
```

```python
#Bir ATM programı yazacaksın. Program kullanıcı çıkış yapana kadar çalışmaya devam etmeli.
```

```python
bakiye = 1000

while True:
    secim = input("Lütfen yapmak istediğiniz işlemi seçin (1/2/3/q): ")
    if secim == "q":
        print("iyi gunler dileriz!")
        break
    elif secim == "1":
        print(f"Bakiyeniz: {bakiye} TL")
        
    elif secim == "2":
        yatirilan_miktar = int(input("Yatırmak istediğiniz miktarı giriniz: "))
        bakiye = bakiye + yatirilan_miktar
        print(f"Yeni bakiyeniz: {bakiye} TL")


    elif secim == "3":
        cekilen_miktar = int(input("Çekmek istediğiniz miktarı giriniz: "))
        if cekilen_miktar > bakiye:
            print("Yetersiz bakiye!! ")
        else:
            bakiye = bakiye - cekilen_miktar
            print(f"Yeni bakiyeniz: {bakiye}")
    else:
        print("Geçersiz seçim, lütfen tekrar deneyiniz! ")
        
        
    



    
```
